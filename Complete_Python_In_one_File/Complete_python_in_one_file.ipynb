```python
rows = int(input("enter number of rows"))
i = 1
k = 1
while i <= rows: 
    # printing spaces
    s = 1
    while s <= rows - i: 
        print(" ", end = " ")
        s += 1

    # printing star
    j = 1
    while j <= k: 
        print("*", end = " ")
        j += 1

    print()
    k += 2
    i += 1
```

## Regular expressions

Regex is a special text strings that allow us to describe search pattern

operation:
1) Find a word in a string
2) Match a single specific letter in a string
3) Match a character in a string
4) match series of range of character in a string
5) replace characters
6) find iterator

applicationi:
1) To validate correct email and phone numbers
2) Also used Web scrapping

character class:
1) [a-z] --> for lowercase characters
2) [A-Z]--> for uppercase characters
3) [0-9]--> for any digits
4) [^a-z]--> for any character not in a range a-z

predifined classes:
'\d'--> for any digits

'\D'--> for any non digits 

'\w' --> for any character (alphanumeric, underscore)

'\W'--> for any non character

'\s'--> for any white spaces in a string

'\S'--> for any Non white spces

common functions:
1) re.findall()--> finds the substring where regular expression matches and returns in the form of list
2) re.search()--> scans through the string , and looks for any location where Regular expression matches
3) re.finditer()--> finds the substring where regular expression matches and returns in the form of iterator
4) re.compile() --> compiles the regular expression pattern for later use
5) re.sub() --> replaces the occurance of regular expression with replacement string

##### Recurssion or recurssive function 

A function which calls itset again and again that is known as recurssive function

```python
# # factorial
# factorial(6) = 6 x 5 x 4 x 3 x 2 x 1
# factorial(5) = 5 x 4 x 3 x 2 x 1
# factorial(4) = 4 x 3 x 2 x 1
# factorial(1) = 1
# factorial(0) = 1
# .
# .
# factorial(n) = n * factorial(n-1)
```

```python
def fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n*fact(n-1)
fact(5)
```

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n * factorial(n-1)
factorial(5)

################
# factorial(5)
# 5 * factorial(4)
# 5 * 4 * factorial(3)
# 5 * 4 * 3 * factorial(2)
# 5 * 4 * 3 * 2 * factorial(1)
# 5 * 4 * 3 * 2 * 1
```

```python
def fact(n,p):
    if p==0:
        return 1
    else:
        return n * fact(n,p-1)
fact(2,3)
```

```python
# sum of n natural numbers.
# eg. for 5 --> 5+4+3+2+1
def add_all(n):
    if n==0:
        return 0
    else:
        return n + add_all(n-1)
add_all(5)
```

```python
# “Reader” can read text
class Reader:
    def read(self, text):
        print(f"{self.__class__.__name__} reads: “{text}”")


# “Writer” can write text
class Writer(Reader):
    def write(self, text):
        print(f"{self.__class__.__name__} writes: “{text}”")


# “Author” does both
class Author(Writer, Reader):
    def create_book(self, title, content):
        print(f"Creating book titled “{title}”")
        self.write(content)


# Usage
a = Author()
a.read("Once upon a time…")  
a.write("Hello, world!")      
a.create_book("My Tale", "It was a dark and stormy night…")
```

#### method overloading

1) Key Points of Method Overloading:
2) Definition: Same method name but different parameters (number or type).
3) Purpose: To perform similar tasks with different inputs.

```python
class MathOperations:
    def add(self, a, b=0, c=0):  # Default values for b and c
        return a + b + c

math_op = MathOperations()
print(math_op.add(15))          # Output: 15 (one argument)
print(math_op.add(5, 30))       # Output: 35 (two arguments)
print(math_op.add(5, 30, 25))   # Output: 60 (three arguments)
```

```python
class MathOperations:
    def add(self, *args):
        result = 0
        for num in args:
            result += num
        return result

math_op = MathOperations()
print(math_op.add(5, 10))       # Output: 15
print(math_op.add(5, 10, 15))   # Output: 30
print(math_op.add(5))           # Output: 5
print(math_op.add())            # Output: 0
```

#### operator overloading

```python
# mathematical operation for vectors
class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __add__(self,other):
        return Vector(self.x+other.x, self.y+other.y)

    def __sub__(self,other):
        return Vector(self.x-other.x, self.y-other.y)

    def __mul__(self,other):
        return Vector(self.x*other, self.y*other)

    def __equal__(self,other):
        return Vector(self.x==other.x, self.y==other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

# create objects of the vector class
v1 = Vector(2,3) # co-ordinates(x,y)
v2 = Vector(4,5) # co-ordinates(x,y) --> this is going to consider as other co-ordinate

print(v1+v2)
print(v1-v2)
print(v1*3)
```

```python
# custom iterator class
class Counter:
    def __init__(self, low, high):
        self.current = low
        self.high = high    # 'high' will be EXCLUDED

    def __iter__(self): # itetor method is used to convert iterable objects into iterator
        return self

    def __next__(self): # next method is used when we want to access any element
        if self.current >= self.high:
            raise StopIteration
        self.current += 1 #  1+1 = 2
        return self.current - 1 # but we have to return 1 first so we are doing 2-1 = 1 and so on...

# Usage (like range)
for i in Counter(1, 4):   # Will print 1, 2, 3 (NOT including 4)
    print(i)
```

#### Generator

A generator is a special type of iterator written using a function and the yield statement, allowing you to declare a function that behaves like an iterator.

Main difference between generator and iterator--> once you accessed an element in generator it is not again accesseble, but in iterator we can access multiple times

```python
# generator concept was introduced so that overall processing ko rok kar ham ek ek value ko lete jaye with the help of yield keyword instead of return 
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

for number in count_up_to(3):
    print(number)      # Output: 1, 2, 3

# yield is similary like return keyword.. but return keword ke call hote hi function ka scope terminate ho jata but, 
# but yield use krne ka faida ye hai ki ham ek ek karke values ko le skte hai..aur function ka scope bhi terminate ni hota
```

```python
# eg. 2

def generator():
    print("Hello World")
    yield 5

    # a = print("Python")
    # return a                  # if used will break scope here

    print("Hello Himanshu")
    yield 10

    print("Hello Arijit")
    yield 15

generator_values = generator()
print(generator_values) # return an generator object

print(next(generator_values))
print(next(generator_values))
print(next(generator_values))
print(next(generator_values)) # jitni bar yield keyword hoga utni hi bar values milegi, otherwise it will so stopIteration error
```
